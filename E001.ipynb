# Exercise: Longest Substring Without Repeating Characters (All Max Cases)
*Objective:*
Given a string, find all substrings of maximum possible length that contain no repeated characters.
Return:
A list of all such substrings (order of appearance in the original string).
The length of these substrings (number of characters in each).

*Return:*
A list of all such substrings (order of appearance in the original string).
The length of these substrings (number of characters in each).

*Details:*
You must consider contiguous substrings only (characters must be consecutive in the original string).
If multiple substrings share the same maximum length, include all of them in the result, preserving their order of occurrence.
If there is only one substring of maximum length, return it in a list (not as a single string).
The second returned value should always be the length of the substring(s), not the count of how many there are.

***

## Examples:

### Example 1
```python
Input:  "abcabcbb"
Output: (["abc", "bca", "cab"], 3)
```
Explanation: The longest substrings without repeating characters are "abc", "bca", and "cab", all of length 3.

### Example 2
```python
Input:  "bbbbbb"
Output: (["b"], 1)
```
Explanation: Only one unique-character substring is possible, "b", of length 1.

### Example 3
```python
Input:  "pwwkew"
Output: (["wke", "kew"], 3)
```
Explanation: The longest substring without repeating characters is "wke", of length 3.

<br>

## Constraints:
Input string length: 0 ≤ len(given_str) ≤ 10⁵<br>
Input may contain spaces, symbols, digits, and letters.<br>
The search is case-sensitive (e.g., "A" and "a" are considered different).<br>

# Final Solution
This is the final solution I developed after much reflection and many iterations. Every step of the code is explained and commented so anyone can not only follow along but also check my reasoning and problem-solving process directly through the code. Variable names and notation are in English to maximize readability and align with Python’s syntax and the exercise context.<br>
I prefer this solution because it’s sharp and reflects my programming style. The sliding-window approach makes it efficient. I also like that it distinguishes between a single correct answer and multiple correct answers, returning either one string or a list of strings as appropriate.

````python
def longestSubS(given_str):
    c = 0 # Counter
    stringi = '' # Working substring
    stringf = '' # Last substring considered as longest
    stringList = [] # List with all the substring with the most characters

    while c < len(given_str):
        # We create a working substring adding characters until one gets repeated
        if given_str[c] not in stringi:
            stringi += given_str[c]
        # Once we find a repeation the working substring is completed:
        else:
            # If we find a substring of the same length as the max (length of stringf), we add it to the list
            if len(stringi) == len(stringf) and not stringi in stringList:
                stringList.append(stringi)
            # If it's the longest so far, reset and update
            if len(stringi) > len(stringf):
                stringf = stringi
                stringList = [stringi]  # Start new list with this substring
            # Adjust the window (remove the first duplicate character)
            stringi += given_str[c] # given_str[c] is the repeated character that led to these action, we'll add it to the working string
            beginat = stringi.find(given_str[c]) + 1
            stringi = stringi[beginat:] # And then chop the working subtring right after the first appearance of that character
        c += 1 # In any case, after each loop we must advance one character down the given string line

    # Final check outside loop (in case the longest substring ends at the last character)
    if len(stringi) == len(stringf) and not stringi in stringList:
        stringList.append(stringi)
    if len(stringi) > len(stringf):
        stringf = stringi
        stringList = [stringi]  # Start new list with this substring

    # Return the result as either a single substring or a list of substrings
    if len(stringList) > 1:
        return stringList, len(stringf)
    return stringf, len(stringf)
````
