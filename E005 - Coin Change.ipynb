# Coin Change

You are given coins of different denominations and a total amount of money amount.
Write a function that computes a combination of coins that adds up to the given amount using the fewest number of coins possible.
<br><br>If it is not possible to make up that amount with any combination of the given coins, return -1.
<br><br>You may assume you have an infinite number of each kind of coin.

***

## Examples <br>

### Example 1:

>Input: coins = [1,5,10], amount = 12<br>
>Output: [10,1,1]<br>
>Explanation: 12 = 10 + 1 + 1. Note that we do not have to use every kind coin available.<br>

### Example 2:

>Input: coins = [2], amount = 3<br>
>Output: -1<br>
>Explanation: The amount of 3 cannot be made up with coins of 2.<br>

<br><br>
***

### Constraints:

1 <= coins.length <= 10 <br>
1 <= coins[i] <= 2^31 - 1 <br>
0 <= amount <= 10000


<br>

***
# Final Solution
The main difficulty here was managing edge cases and bugs that could disrupt the logic, while keeping the code clean and letting the solution naturally reach the correct output instead of depending on hard-coded exceptions

```python
def minCoins(coins, num):

    ''' DECLARING GLOBAL VARIABLES '''
    i = 0
    routes = [[]]
    newRoutes = []
    modifications = False

    ''' IF NEEDED WE ADD CYCLICALY ONE OF EACH COIN TO EACH ROUTE WITHOUT REPETITIONS '''
    while i < len(routes):
        totalAmount = sum(routes[i])
        if totalAmount == num:
            return routes[i]
        elif totalAmount < num:
            for coin in coins:
                workingRoute = routes[i].copy()
                workingRoute.append(coin)
                workingRoute.sort(reverse=True)
                if sum(workingRoute) == num:
                    return workingRoute
                elif (sum(workingRoute) < num) and workingRoute not in newRoutes:
                    newRoutes.append(workingRoute)
                    modifications = True

        ''' FIRST EXECUTION '''
        if routes == [[]]: # Updating list first time around
            routes = newRoutes.copy()
            newRoutes = []
            if not routes or all(x == [0] for x in routes): # Handeling empty list
                return -1
            i -= 1

        ''' END OF THE ROAD INSTRUCTIONS '''
        if i == len(routes)-1:
            routes = newRoutes.copy() # UPDATE ROUTES
            newRoutes = []
            i = 0 # CYCLING MECHANISM
            if not modifications: # BREAKING MECHANISM
                return -1
            modifications = False
        else:
            i += 1
```

## Edge Cases

```python
# IMPOSSIBLE TO MAKE UP THE AMOUNT
minCoins([8,4],25)
```

```python
# COIN = 0
minCoins([0],45)
```

```python
# NO COINS
minCoins([],20)
```

```python
# ALL COINS > NUM
minCoins([10,20,50],2)
```

```python
# ONE COIN == NUM
minCoins([10,20,50],20)
```

```python
# AMOUNT = 0
minCoins([10,5,2],0)
```
