# Coin Change

You are given coins of different denominations and a total amount of money amount. Write a function to compute the fewest number of coins that you need to make up that amount. If that amount of money cannot be made up by any combination of the coins, return -1.

You may assume that you have an infinite number of each kind of coin.

Example 1:

Input: coins = [1,5,10], amount = 12

Output: 3
Explanation: 12 = 10 + 1 + 1. Note that we do not have to use every kind coin available.

Example 2:

Input: coins = [2], amount = 3

Output: -1
Explanation: The amount of 3 cannot be made up with coins of 2.

Example 3:

Input: coins = [1], amount = 0

Output: 0
Explanation: Choosing 0 coins is a valid way to make up 0.

Constraints:

1 <= coins.length <= 10
1 <= coins[i] <= 2^31 - 1
0 <= amount <= 10000

```python
c = [2,5,10]
n = 26
```

```python
def minCoins(coins, num):

    ''' CREATING THE ROUTES LIST '''
    coins.sort(reverse=True)
    routes = []
    for coin in coins:
        routes.append([coin])

    ''' DECLARING GLOBAL VARIABLES '''
    i = 0
    newRoutes = []
    modifications = False

    ''' IF NEEDED WE ADD CYCLICALY ONE OF EACH COIN TO EACH ROUTE WITHOUT REPEATITIONS '''
    while i < len(routes):
        totalAmount = sum(routes[i])
        if totalAmount == num:
            return routes[i]
        if totalAmount < num:
            for coin in coins:
                workingRoute = routes[i].copy()
                workingRoute.append(coin)
                workingRoute.sort(reverse=True)
                if sum(workingRoute) == num:
                    return workingRoute
                elif (sum(workingRoute) < num) and workingRoute not in newRoutes:
                    newRoutes.append(workingRoute)
                    modifications = True
            
            i += 1
        
        ''' END OF THE ROAD INSTRUCTIONS '''
        if i == len(routes)-1:
            routes = newRoutes # UPDATE ROUTES
            i = 0 # CYCLING MECHANISM
            if not modifications: # BREAKING MECHANISM
                return -1
            
        modifications = False
```

```python
def minCoinsDOS(coins, num):

    ''' DECLARING GLOBAL VARIABLES '''
    i = 0
    routes = [[]]
    newRoutes = []
    modifications = False

    ''' IF NEEDED WE ADD CYCLICALY ONE OF EACH COIN TO EACH ROUTE WITHOUT REPEATITIONS '''
    while i < len(routes):
        totalAmount = sum(routes[i])
        if totalAmount == num:
            return routes[i]
        elif totalAmount < num:
            for coin in coins:
                workingRoute = routes[i].copy()
                workingRoute.append(coin)
                workingRoute.sort(reverse=True)
                if sum(workingRoute) == num:
                    return workingRoute
                elif (sum(workingRoute) < num) and workingRoute not in newRoutes:
                    newRoutes.append(workingRoute)
                    modifications = True
            
        i += 1

        ''' FIRST EXECUTION '''
        if routes == [[]]: # Updating list first time around
            routes = newRoutes
            if not routes or all(x == [0] for x in routes): # Handeling empty list
                return -1

        ''' END OF THE ROAD INSTRUCTIONS '''
        if i == len(routes)-1:
            routes = newRoutes # UPDATE ROUTES
            i = 0 # CYCLING MECHANISM
            if not modifications: # BREAKING MECHANISM
                return -1
            
        modifications = False
```
