```python
c = [2,5,10]
n = 26
```

```python
def minCoins(coins, num):

    # CREATING THE ROUTE LISTS
    coins.sort(reverse=True)
    routes = []
    for coin in coins:
        routes.append([coin])

    i = 0
    newRoutes = []

    while i < len(routes):
        totalAmount = sum(routes[i])
        if totalAmount == num:
            return routes[i]
        if totalAmount < num:
            for coin in coins:
                workingRoute = routes[i].copy()
                workingRoute.append(coin)
                workingRoute.sort(reverse=True)
                if sum(workingRoute) == num:
                    return workingRoute
                elif (sum(workingRoute) < num) and workingRoute not in newRoutes:
                    newRoutes.append(workingRoute)
            
            i += 1
        
        if i == len(routes)-1:
            routes = newRoutes
            i = 0
```

```python
minCoins(c, n)
```
