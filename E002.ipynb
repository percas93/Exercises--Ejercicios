```python
list1 = [2,8,9,11,14,6,4,3]
```

```python
def mostLiquid(givenList):
    
    maxArea = 0
    finalResult = None
    
    for an, a in enumerate(givenList): # For each element on the list..
        w = an + 1
        for bn, b in enumerate(givenList[w:], start=w): # ..we calculate the area it creates with the next ones
            width = bn - an # The distance between them give us the width
            height = min(a,b)
            area = width * height
            if area > maxArea:
                maxArea = area # The highest result gets save to compare and find the max area
                finalResult = an,bn # But what we want are its coordanades

            
    return(finalResult)
```

```python
mostLiquid(list1)
```

```python
mostLiquid([1])
```
