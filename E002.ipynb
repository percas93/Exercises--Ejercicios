# Container With Most Water

<br>

You are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the ith line are (i, 0) and (i, height[i]).

Find two lines that together with the x-axis form a container, such that the container contains the most water.

Return the maximum amount of water a container can store.

Notice that you may not slant the container.


<br><br>

***

## Example 1:

<br>

![Example 1 Image](FILES/Container%20With%20Most%20Water.jpg "Example 1")

<br>
Input: height = [1,8,6,2,5,4,8,3,7]<br>
Output: 49<br>
Explanation: The above vertical lines are represented by array [1,8,6,2,5,4,8,3,7]. In this case, the max area of water (blue section) the container can contain is 49.<br>

<br>

Example 2:

Input: height = [1,1]
Output: 1
 
<br>

## Constraints:

n == height.length
2 <= n <= 105
0 <= height[i] <= 104


<br><br>

# Final Solution

This is my very personal solution to solve the problem as indicated in the exercise. I like it because is clean, redable and scalable.<br>

>Further down you will find a even more interesting solution that applies this function to create a visual representation of the "water container" and coordinates ("walls"). This emulating the example image given by the exercise statement. '''

```python
def mostLiquid(givenList):
    
    maxArea = 0
    #finalResult = None
    
    for an, a in enumerate(givenList): # For each element on the list..
        w = an + 1
        for bn, b in enumerate(givenList[w:], start=w): # ..we calculate the area it creates with the next ones
            width = bn - an # The distance between them give us the width
            height = min(a,b)
            area = width * height
            if area > maxArea:
                maxArea = area # The highest result gets save to compare and find the max area
                #finalResult = an,bn # For me its even more intersting to get its coordanades

    #return(finalResult)        
    return(maxArea)
```
